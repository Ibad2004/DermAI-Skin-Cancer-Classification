```python
import matplotlib.pyplot as plt

# Models and accuracies
models = ["Baseline CNN", "ResNet50"]
accuracies = [60.68, 74.65]

plt.figure(figsize=(8,5))

bars = plt.bar(
    models,
    accuracies
)

# Add values above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.5,
        f"{height:.2f}%",
        ha='center',
        fontsize=11
    )

plt.ylabel("Test Accuracy (%)")
plt.xlabel("Models")
plt.title("Test Accuracy Comparison")
plt.ylim(0,100)

plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Models and Weighted F1 scores
models = ["Baseline CNN", "ResNet50"]
f1_scores = [0.63, 0.77]

plt.figure(figsize=(8,5))

bars = plt.bar(
    models,
    f1_scores
)

# Add values above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.01,
        f"{height:.2f}",
        ha='center',
        fontsize=11
    )

plt.ylabel("Weighted F1 Score")
plt.xlabel("Models")
plt.title("Weighted F1 Score Comparison")
plt.ylim(0,1)

plt.grid(axis='y', linestyle='--', alpha=0.5)

# Optional save
plt.savefig(
    "../reports/figures/weighted_f1_comparison.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
